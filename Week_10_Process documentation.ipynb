## 1.1 Problem 1: Tip Calculator

### 1.1.1 The Scenario

You work at a Perth restaurant. The manager wants a simple tool where staff can enter:

- The bill amount (e.g., \$45.80)
- The tip percentage (e.g., 15%)

And get back:

- The tip amount
- The total amount to charge

---

### 1.1.2 What You Must Submit

## STEP 1: Understand the Problem

In your own words, what is this program supposed to do? (2-3 sentences)

**Ans:**
      This program is supposed to help restaurant staff calculate tips and total bill, quickly and accurately. The user enters the customer’s bill amount and the percentage they want to tip, and then the program calculates the tip amount and adds it to the bill to show the total amount the customer needs to pay.

---  

## STEP 2: Identify Inputs and Outputs

### INPUTS:

- What does the user provide?

**Ans:**
The user provides two input that is the bill amount and the tip percentage. The bill amount is basically the total cost of the customer’s food and drinks before adding the tip, and the tip percentage is the percentage of extra money the customer wants to give as a tip for the service.

- What type of data? Numbers? Text?

**Ans:**
The input data are numbers (decimal and percentage values).Specifically , the bill amount (e.g., \$45.80) is a decimal number because it includes cents, and the tip percentage  (e.g., 15%) is also a number that will later be converted into a decimal for calculation purposes.  

### OUTPUTS:

- What does the program show back?  
**Ans:**
The program shows back the tip amount in dollars and the final total amount the customer must pay after the tip has been added to the original bill.

- What format?  
**Ans:** The results should be displayed as currency values meaning it should be displayed in dollar format "\$" usually with 2 decimal places.

---

## STEP 3: Work It By Hand

Show me you can do this with a calculator. Use this example:

- Bill amount: \$45.80  
- Tip percentage: 15%

**Ans:** 

The steps:

1. First calculation:  
   I would convert the tip percentage into a decimal by dividing 15 by 100, which gives 0.15.  
   15% = 15/100 = 0.15  

2. Second calculation:  
   I would multiply the bill amount by the decimal tip percentage:  
   45.80 × 0.15=  \$6.87  

4. Third step:  
   After finding the tip amount, I would add the tip amount to the original bill amount:  
   \$45.80 + \$6.87 =
                      \$52.67  

6. Finally, I would tell the staff member that:  
   “The tip amount is 6.87 and the total charge is \$52.67.”

## 1.2 Problem 2: Late Fee Calculator

### 1.2.1 The Scenario

You work for a library in Perth. Books are due back in 14 days. For every day late, there’s a \$0.50 fine

(maximum \$10.00).

The librarian wants a program where they enter:

- How many days the book is overdue

And get back:

- The fine amount (or “Maximum fine reached” if it’s over \$10)

---

### 1.2.2 What You Must Submit

## STEP 1: Understand the Problem

**Ans:**
This program is designed to help librarians in Perth calculate overdue fines for library books. The librarian enters the number of days the book is overdue, and the program calculates the fine based on the library’s rule of charging 0.50 dollars for each late day.The program must also make sure that the fine does not go above the maximum limit of 10.00. If the calculated fine is higher than 10.00, the program should display that the maximum fine has been reached.

---

## STEP 2: Identify Inputs and Outputs

### INPUTS:

- What information does the librarian provide?  

**Ans:** 
The librarian provides the number of days the book is overdue. This is a whole number because days are counted in full numbers. This input is used to calculate the total fine.

### OUTPUTS:

- What does the program calculate and show?*  

**Ans:**
The program calculates and shows the overdue fine amount based on the number of days the book is late. The fine is calculated by multiplying the overdue days by $0.50 per day.

- Any special cases?  

 **Ans:**
If the fine goes above 10.00 dollars , the program should display “Maximum fine reached” and charge only the maximum amount of 10.00 dollars.
If the book is returned on time (0 days late), the program displays a fine of \$0.00 because there is no charge.

---

## STEP 3: Work It By Hand

### Case A: Book is 5 days late

- My calculation:  
  5 × \$0.50 = \$2.50

- Fine:  
  \$2.50

---

### Case B: Book is 25 days late

- My calculation:  
  25 × \$0.50 = \$12.50

- Fine:  
  \$10.00

- Special note:  
  The calculated fine is more than the maximum allowed fine, so the fine is capped at \$10.00.

---

### Case C: Book is 0 days late

-My calculation:
0 × \$0.50 = \$0.00

-Fine:  
  \$0.00
