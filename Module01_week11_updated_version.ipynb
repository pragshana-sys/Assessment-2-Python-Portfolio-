### **Scenario**: I need a function that converts a temperature value from Celsius to Fahrenheit.

# **Specification:**

This function must convert a temperature value given in Celsius (°C) into its equivalent value in Fahrenheit (°F).

The temperature conversion function must:

•	Take a numeric input representing temperature in Celsius (°C).

•	Convert it to Fahrenheit using the standard conversion formula: 
      °F = (°C × 9/5) + 32
      
•	Return the converted Fahrenheit value 


**Examples:**  
•	0°C → 32°F (C=0)   {0°C = (0 × 9/5) + 32 = 32°F  
•	100°C → 212°F (boiling point of water)   
•	-40°C → -40°F (same value in both scales C = F) 



**Edge cases:**

•	Extreme low temperature (absolute limit case):
-273.15°C = -459.67°F
(This is approximately absolute zero, the lowest possible temperature in physics.)   

•°F = °C (-40 °C = -40 °F) when °Celsius equals to °Fahrenheit 

•	Very large value:
Input: 1,000,000°C → Output: 1,800,032°F

# 1. Implementation with the help of AI 

**My AI prompt:**  
Help me create a beginner-friendly Python program that converts Celsius to Fahrenheit using the formula F = (C × 9/5) + 32. The program should ask the user for input and display the result.   



**AI reply code:**

````python
def celsius_to_fahrenheit(celsius):
    return (celsius * 9/5) + 32

# Ask user for input
celsius = float(input("Enter temperature in Celsius: "))

# Convert and display result
fahrenheit = celsius_to_fahrenheit(celsius)

print("Temperature in Fahrenheit:", fahrenheit)
````

#  Code:

````python
def celsius_to_fahrenheit(celsius):
    return (celsius * 9/5) + 32


try:
    # Ask user for input
    celsius = float(input("Enter temperature in Celsius: "))

    # Convert temperature
    fahrenheit = celsius_to_fahrenheit(celsius)

    # Display result
    print("Temperature in Fahrenheit:", fahrenheit)

except ValueError:
    # Handle invalid input
    print("Invalid input! Please enter a numeric value.")


# Assertions (tests)
assert celsius_to_fahrenheit(0) == 32
assert celsius_to_fahrenheit(100) == 212
assert celsius_to_fahrenheit(-40) == -40
assert round(celsius_to_fahrenheit(-273.15), 2) == -459.67
assert celsius_to_fahrenheit(1000000) == 1800032

print("\n All Assert tests passed.")
````

### Scenario

### Requirements
A strong password must:
- Be at least 8 characters long
- Include one uppercase letter
- Include one number

### Reflection

````python
# ==========================================
# Module 1 – Practice Challenge

# ------------------------------------------
# Challenge  – Addition Function
# ------------------------------------------

def add(a, b):
    return a + b


# Assertions
assert add(2, 3) == 5
assert add(-1, 1) == 0
assert add(0, 0) == 0

print("Addition function tests passed.")


# ------------------------------------------
# Challenge – Discount Calculator
# ------------------------------------------

def calculate_discount(price, percent):
    return price * (percent / 100)


# Assertions
assert calculate_discount(100, 10) == 10
assert calculate_discount(50, 20) == 10
assert calculate_discount(200, 5) == 10

print("Discount calculator tests passed.")


# ------------------------------------------
# Challenge – Password Strength Checker
# ------------------------------------------

def is_strong_password(password):

    # Password must be at least 8 characters
    if len(password) < 8:
        return False

    has_number = False
    has_uppercase = False

    # Check characters
    for char in password:

        if char.isdigit():
            has_number = True

        if char.isupper():
            has_uppercase = True

    return has_number and has_uppercase


# Assertions
assert is_strong_password("Pass123") == False
assert is_strong_password("password1") == False
assert is_strong_password("Password1") == True
assert is_strong_password("PASSWORD1") == True
assert is_strong_password("") == False
assert is_strong_password("Pass") == False

print("Password strength tests passed.")


print("\nAll Module 1 practice challenge tests passed successfully.")
````
